## Breast Cancer Dataset

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import r2_score

# LIME classique (uniquement pour LinearRegression)
from lime.lime_tabular import LimeTabularExplainer as LimeClassicExplainer
# LIME-NDT
from lime_ndt.lime_tabular import LimeNdtExplainer as LimeNDTExplainer
from lime_ndt.utils.ndt_sklearn_wrapper import NDTRegressorWrapper

# ========================
# Wrapper pour DecisionTree
# ========================
class DecisionTreeWrapper(DecisionTreeRegressor):
    def fit(self, X, y, sample_weight=None, *args, **kwargs):
        super().fit(X, y, sample_weight=sample_weight, *args, **kwargs)
        self.coef_ = np.array(self.feature_importances_)
        self.intercept_ = 0
        return self

# ========================
# Charger dataset
# ========================
data = load_breast_cancer()
X = data.data
y = data.target.astype(float)  # convert to float pour la régression
feature_names = data.feature_names
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
X_train, X_test = np.asarray(X_train, dtype=float), np.asarray(X_test, dtype=float)

# ========================
# Explainers
# ========================
explainer_classic = LimeClassicExplainer(
    X_train,
    feature_names=feature_names,
    discretize_continuous=False,
    mode='classification'  # pour LinearRegression local
)
explainer_ndt = LimeNDTExplainer(
    X_train,
    feature_names=feature_names,
    discretize_continuous=False,
    mode='classification'
)

# ========================
# Fonction fidélité locale
# ========================
def fidelity_local_model(instance, explainer, local_model, predict_fn,
                         num_samples_train=5000, num_samples_test=500, metric="r2"):
    num_features = instance.shape[0]

    # --- Train perturbations ---
    Z_train = np.zeros((num_samples_train, num_features))
    for i in range(num_features):
        mean = instance[i]
        std = explainer.scaler.scale_[i] if hasattr(explainer.scaler, "scale_") else 0.01
        Z_train[:, i] = np.random.normal(mean, std, size=num_samples_train)
    y_global_train = predict_fn(Z_train)
    local_model.fit(Z_train, y_global_train)

    # --- Test perturbations ---
    Z_test = np.zeros((num_samples_test, num_features))
    for i in range(num_features):
        mean = instance[i]
        std = explainer.scaler.scale_[i] if hasattr(explainer.scaler, "scale_") else 0.01
        Z_test[:, i] = np.random.normal(mean, std, size=num_samples_test)
    y_global_test = predict_fn(Z_test)
    y_local_test = local_model.predict(Z_test)

    if metric == "r2":
        return r2_score(y_global_test, y_local_test)
    elif metric == "mse":
        from sklearn.metrics import mean_squared_error
        return mean_squared_error(y_global_test, y_local_test)

# ========================
# Variation de la complexité du réseau global
# ========================
hidden_layer_configs = [
    (),                            # baseline : LinearRegression-like (pas de couche)
    (64,),                         # 1 couche large
    (128,64),                       # 2 couches
    (256,128,64),                   # 3 couches
    (256,256,128,64),               # 4 couches
    (512,256,128,64),               # 5 couches
    (512,512,256,128,64,32)         # 6 couches (très gros réseau)
]
# Profondeur de l'arbre proportionnelle à la complexité
tree_depths = [1, 2, 3, 4, 5, 6, 7]

results_complexity = {"LinearRegression": [], "DecisionTree": [], "NDT": []}
complexity_axis = []

for i, hl in enumerate(hidden_layer_configs):
    print(f"==> Complexité {i}: {hl if hl else 'Linear'}")

    # Modèle global
    if hl:
        mlp_global = MLPClassifier(
            hidden_layer_sizes=hl,
            activation='relu',
            solver='adam',
            max_iter=4000,        # ⬆️ plus d'itérations pour converger
            random_state=42
        )
    else:
        # Baseline quasi-linéaire
        mlp_global = MLPClassifier(
            hidden_layer_sizes=(1,),
            activation='identity',
            solver='adam',
            max_iter=2000,
            random_state=42
        )

    mlp_global.fit(X_train, y_train)

    def predict_fn(X):
        return mlp_global.predict(X).astype(float)

    # Fidelity
    scores_lr, scores_dt, scores_ndt = [], [], []
    for idx in range(3):  # quelques instances pour la moyenne
        instance = X_test[idx]

        scores_lr.append(
            fidelity_local_model(instance, explainer_classic,
                                 LinearRegression(), predict_fn)
        )

        depth = tree_depths[min(i, len(tree_depths)-1)]
        scores_dt.append(
            fidelity_local_model(instance, explainer_ndt,
                                 DecisionTreeWrapper(max_depth=depth), predict_fn)
        )

        scores_ndt.append(
            fidelity_local_model(instance, explainer_ndt,
                                 NDTRegressorWrapper(D=X_train.shape[1], gammas=[1,1]),
                                 predict_fn)
        )

    results_complexity["LinearRegression"].append(np.mean(scores_lr))
    results_complexity["DecisionTree"].append(np.mean(scores_dt))
    results_complexity["NDT"].append(np.mean(scores_ndt))
    complexity_axis.append(i)

# ========================
# Tracé
# ========================
plt.figure(figsize=(10,6))
plt.plot(complexity_axis, results_complexity["LinearRegression"], 'o-', label="Linear Regression")
plt.plot(complexity_axis, results_complexity["DecisionTree"], 's-', label="Decision Tree")
plt.plot(complexity_axis, results_complexity["NDT"], 'd-', label="Neural Decision Tree")

plt.xticks(complexity_axis, ["LR","1","2","3","4","5","6"])
plt.xlabel("Complexité du modèle (nb de couches cachées)")
plt.ylabel("Fidélité (R²)")
plt.title("Fidélité vs Complexité du modèle (Breast Cancer Dataset)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_covtype
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import r2_score

# LIME classique (uniquement pour LinearRegression)
from lime.lime_tabular import LimeTabularExplainer as LimeClassicExplainer
# LIME-NDT
from lime_ndt.lime_tabular import LimeNdtExplainer as LimeNDTExplainer
from lime_ndt.utils.ndt_sklearn_wrapper import NDTRegressorWrapper

# ========================
# Wrapper pour DecisionTree
# ========================
class DecisionTreeWrapper(DecisionTreeRegressor):
    def fit(self, X, y, sample_weight=None, *args, **kwargs):
        super().fit(X, y, sample_weight=sample_weight, *args, **kwargs)
        self.coef_ = np.array(self.feature_importances_)
        self.intercept_ = 0
        return self

# ========================
# Charger dataset
# ========================
data = fetch_covtype()
data = fetch_covtype()
X = data.data[:2000]
y = data.target[:2000].astype(float)  # convert to float pour la régression
feature_names = data.feature_names
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
X_train, X_test = np.asarray(X_train, dtype=float), np.asarray(X_test, dtype=float)

# ========================
# Explainers
# ========================
explainer_classic = LimeClassicExplainer(
    X_train,
    feature_names=feature_names,
    discretize_continuous=False,
    mode='classification'  # pour LinearRegression local
)
explainer_ndt = LimeNDTExplainer(
    X_train,
    feature_names=feature_names,
    discretize_continuous=False,
    mode='classification'
)

# ========================
# Fonction fidélité locale
# ========================
def fidelity_local_model(instance, explainer, local_model, predict_fn,
                         num_samples_train=5000, num_samples_test=500, metric="r2"):
    num_features = instance.shape[0]

    # --- Train perturbations ---
    Z_train = np.zeros((num_samples_train, num_features))
    for i in range(num_features):
        mean = instance[i]
        std = explainer.scaler.scale_[i] if hasattr(explainer.scaler, "scale_") else 0.01
        Z_train[:, i] = np.random.normal(mean, std, size=num_samples_train)
    y_global_train = predict_fn(Z_train)
    local_model.fit(Z_train, y_global_train)

    # --- Test perturbations ---
    Z_test = np.zeros((num_samples_test, num_features))
    for i in range(num_features):
        mean = instance[i]
        std = explainer.scaler.scale_[i] if hasattr(explainer.scaler, "scale_") else 0.01
        Z_test[:, i] = np.random.normal(mean, std, size=num_samples_test)
    y_global_test = predict_fn(Z_test)
    y_local_test = local_model.predict(Z_test)

    if metric == "r2":
        return r2_score(y_global_test, y_local_test)
    elif metric == "mse":
        from sklearn.metrics import mean_squared_error
        return mean_squared_error(y_global_test, y_local_test)

# ========================
# Variation de la complexité du réseau global
# ========================
hidden_layer_configs = [
    (),                            # baseline : LinearRegression-like (pas de couche)
    (64,),                         # 1 couche large
    (128,64),                       # 2 couches
    (256,128,64),                   # 3 couches
    (256,256,128,64),               # 4 couches
    (512,256,128,64),               # 5 couches
    (512,512,256,128,64,32)         # 6 couches (très gros réseau)
]
# Profondeur de l'arbre proportionnelle à la complexité
tree_depths = [1, 2, 3, 4, 5, 6, 7]

results_complexity = {"LinearRegression": [], "DecisionTree": [], "NDT": []}
complexity_axis = []

for i, hl in enumerate(hidden_layer_configs):
    print(f"==> Complexité {i}: {hl if hl else 'Linear'}")

    # Modèle global
    if hl:
        mlp_global = MLPClassifier(
            hidden_layer_sizes=hl,
            activation='relu',
            solver='adam',
            max_iter=4000,        # ⬆️ plus d'itérations pour converger
            random_state=42
        )
    else:
        # Baseline quasi-linéaire
        mlp_global = MLPClassifier(
            hidden_layer_sizes=(1,),
            activation='identity',
            solver='adam',
            max_iter=2000,
            random_state=42
        )

    mlp_global.fit(X_train, y_train)

    def predict_fn(X):
        return mlp_global.predict(X).astype(float)

    # Fidelity
    scores_lr, scores_dt, scores_ndt = [], [], []
    for idx in range(3):  # quelques instances pour la moyenne
        instance = X_test[idx]

        scores_lr.append(
            fidelity_local_model(instance, explainer_classic,
                                 LinearRegression(), predict_fn)
        )

        depth = tree_depths[min(i, len(tree_depths)-1)]
        scores_dt.append(
            fidelity_local_model(instance, explainer_ndt,
                                 DecisionTreeWrapper(max_depth=depth), predict_fn)
        )

        scores_ndt.append(
            fidelity_local_model(instance, explainer_ndt,
                                 NDTRegressorWrapper(D=X_train.shape[1], gammas=[1,1]),
                                 predict_fn)
        )

    results_complexity["LinearRegression"].append(np.mean(scores_lr))
    results_complexity["DecisionTree"].append(np.mean(scores_dt))
    results_complexity["NDT"].append(np.mean(scores_ndt))
    complexity_axis.append(i)

# ========================
# Tracé
# ========================
plt.figure(figsize=(10,6))
plt.plot(complexity_axis, results_complexity["LinearRegression"], 'o-', label="Linear Regression")
plt.plot(complexity_axis, results_complexity["DecisionTree"], 's-', label="Decision Tree")
plt.plot(complexity_axis, results_complexity["NDT"], 'd-', label="Neural Decision Tree")

plt.xticks(complexity_axis, ["LR","1","2","3","4","5","6"])
plt.xlabel("Complexité du modèle (nb de couches cachées)")
plt.ylabel("Fidélité (R²)")
plt.title("Fidélité vs Complexité du modèle (Covtype Dataset)")
plt.legend()
plt.grid(True)
plt.show()
```
